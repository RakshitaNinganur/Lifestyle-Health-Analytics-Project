```python
correlation = df.select_dtypes(include="number").corr()

plt.figure(figsize=(10,6))

plt.imshow(correlation)

plt.colorbar()

plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=90)

plt.yticks(range(len(correlation.columns)), correlation.columns)

plt.title("Correlation Heatmap")

plt.show()
```

```python
health_gender = df.groupby("Gender")["Health_Score"].mean()

plt.figure(figsize=(6,4))

plt.bar(health_gender.index, health_gender.values)

plt.title("Average Health Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Health Score")

plt.show()
```

```python
avg_bmi = df.groupby("Age_group")["BMI"].mean()

plt.figure(figsize=(8,5))

plt.bar(avg_bmi.index, avg_bmi.values)

plt.title("Average BMI by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Average BMI")

plt.show()
```

```python
disease_risk = df["Disease_Risk"].value_counts()

plt.figure(figsize=(7,5))

plt.bar(disease_risk.index, disease_risk.values)

plt.title("Disease Risk Distribution")
plt.xlabel("Disease Risk")
plt.ylabel("Number of People")

plt.show()
```

```python
sleep_occupation = df.groupby("Occupation")["Sleep_Hours"].mean()

plt.figure(figsize=(10,5))

plt.bar(sleep_occupation.index, sleep_occupation.values)

plt.title("Average Sleep Hours by Occupation")
plt.xlabel("Occupation")
plt.ylabel("Average Sleep Hours")

plt.xticks(rotation=45)

plt.show()
```

```python
stress_occupation = df.groupby("Occupation")["Stress_Level"].mean()

plt.figure(figsize=(10,5))

plt.bar(stress_occupation.index, stress_occupation.values)

plt.title("Average Stress Level by Occupation")
plt.xlabel("Occupation")
plt.ylabel("Average Stress Level")

plt.xticks(rotation=45)

plt.show()
```

```python
health_bmi = df.groupby("BMI_Category")["Health_Score"].mean()

plt.figure(figsize=(8,5))

plt.bar(health_bmi.index, health_bmi.values)

plt.title("Average Health Score by BMI Category")
plt.xlabel("BMI Category")
plt.ylabel("Average Health Score")

plt.show()
```

```python
water_risk = df.groupby("Disease_Risk")["Water_Intake_L"].mean()

plt.figure(figsize=(7,5))

plt.bar(water_risk.index, water_risk.values)

plt.title("Average Water Intake by Disease Risk")
plt.xlabel("Disease Risk")
plt.ylabel("Average Water Intake (Liters)")

plt.show()
```

```python
city_health = df.groupby("City")["Health_Score"].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))

plt.bar(city_health.index, city_health.values)

plt.title("Top 10 Cities by Average Health Score")
plt.xlabel("City")
plt.ylabel("Average Health Score")

plt.xticks(rotation=45)

plt.show()
```

```python
cholesterol_age = df.groupby("Age_group")["Cholesterol"].mean()

plt.figure(figsize=(8,5))

plt.bar(cholesterol_age.index, cholesterol_age.values)

plt.title("Average Cholesterol by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Average Cholesterol")

plt.show()
```

```python
blood_sugar_bmi = df.groupby("BMI_Category")["Blood_Sugar"].mean()

plt.figure(figsize=(8,5))

plt.bar(blood_sugar_bmi.index, blood_sugar_bmi.values)

plt.title("Average Blood Sugar by BMI Category")
plt.xlabel("BMI Category")
plt.ylabel("Average Blood Sugar")

plt.show()
```

```python
smoking_risk = pd.crosstab(df["Smoking_Status"], df["Disease_Risk"])

smoking_risk.plot(kind="bar", figsize=(8,5))

plt.title("Smoking Status vs Disease Risk")
plt.xlabel("Smoking Status")
plt.ylabel("Number of People")

plt.xticks(rotation=0)

plt.show()
```

```python
alcohol_health = df.groupby("Alcohol_Consumption")["Health_Score"].mean()

plt.figure(figsize=(8,5))

plt.bar(alcohol_health.index, alcohol_health.values)

plt.title("Average Health Score by Alcohol Consumption")
plt.xlabel("Alcohol Consumption")
plt.ylabel("Average Health Score")

plt.show()
```
